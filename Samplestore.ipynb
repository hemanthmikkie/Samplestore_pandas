```python
data_clean=data.copy()
```

```python
data_clean["Order Date"] = pd.to_datetime(
    data_clean["Order Date"],
    errors="coerce"
)

data_clean["Ship Date"] = pd.to_datetime(
    data_clean["Ship Date"],
    errors="coerce"
)
```

```python
txt_col=data_clean.select_dtypes(include="object").columns
txt_col
for col in txt_col:
    data_clean[col]=data_clean[col].str.strip() #strip() removes spaces from the beginning and end of strings.
```

```python
total_sales = data_clean["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
total_profit = data_clean["Profit"].sum()
print("Total Profit:", total_profit)
```

```python
total_quantity = data_clean["Quantity"].sum()
print("Total Quantity:", total_quantity)
```

```python
total_orders = data_clean["Order ID"].nunique()
print("Total Orders:", total_orders)
```

```python
total_customers = data_clean["Customer ID"].nunique()
print("Total Customers:", total_customers)
```

```python
profit_margin = (
    total_profit / total_sales
) * 100 if total_sales != 0 else np.nan

print("Profit Margin:", profit_margin)
```

```python
kpi_report = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Total Orders",
        "Total Customers",
        "Profit Margin (%)"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        total_orders,
        total_customers,
        profit_margin
    ]
})

kpi_report
```

```python
category_sales= (
    data_clean.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_sales.plot(
    kind="bar",
    title="Sales by Category",
    xlabel="Categoty",
    ylabel="Sales")
plt.xticks(rotation=0)
plt.show()
    
```

```python
category_profit= (
    data_clean.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

category_profit
```

```python
category_profit.plot(
    kind="bar",
    title="Profit by Category",
    # xlabel="Category",
    # ylabel="Profit",
color="green")
plt.ylabel("Profit")
plt.show()
```

```python
subcategory_analysis= (
    data_clean.groupby("Sub-Category")
    .agg({
        "Sales": "sum",
        "Profit": "sum",
        "Quantity": "sum"
    })
    .sort_values("Sales", ascending=False)
)

subcategory_analysis
```

```python
subcategory_analysis.sort_values(
    "Profit"
).head(5)
```

```python
subcategory_analysis["Sales"].sort_values().plot(
    kind="barh",
    title="Sales by Subcategory"
)

plt.xlabel("Sales")
plt.show()
```

```python
region_analysis= (
    data_clean.groupby("Region")
    .agg({
        "Sales": "sum",
        "Profit": "sum",
        "Quantity": "sum",
        "Order ID": "nunique"
    })
    .rename(columns={
        "Order ID": "Total Orders"
    })
)

region_analysis
```

```python
region_analysis["Sales"].plot(
    kind="bar",
    title="Sales by Region"
)

plt.ylabel("Sales")
plt.show()
```

```python
region_analysis["Profit"].plot(
    kind="bar",
    title="Profit by Region"
)

plt.ylabel("Profit")
plt.show()
```

```python
segment_analysis = (
    data_clean.groupby("Segment")
    .agg({
        "Sales": "sum",
        "Profit": "sum",
        "Customer ID": "nunique",
        "Order ID": "nunique"
    })
    .rename(columns={
        "Customer ID": "Customers",
        "Order ID": "Orders"
    })
)

segment_analysis
```

```python
segment_analysis["Sales"].plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Sales Contribution by Segment"
)

plt.ylabel("")
plt.show()
```

```python
data_clean["Year"] = data_clean["Order Date"].dt.year

data_clean["Month"] = data_clean["Order Date"].dt.month

data_clean["Month_Name"] = (
    data_clean["Order Date"].dt.month_name()
)
```

```python
monthly_sales = (
    data_clean.groupby(
        data_clean["Order Date"].dt.to_period("M")
    )["Sales"]
    .sum()
)

monthly_sales
```

```python
monthly_sales.plot(
    kind="line",
    marker="o",
    title="Monthly Sales Trend"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
yearly_sales = (
    data_clean.groupby("Year")["Sales"]
    .sum()
)

yearly_sales
```

```python
yearly_sales.plot(
    kind="bar",
    title="Yearly Sales Performance"
)

plt.ylabel("Sales")
plt.show()
```

```python
data_clean["Shipping Days"] = (
    data_clean["Ship Date"] -
    data_clean["Order Date"]
).dt.days
```

```python
shipping_analysis = (
    data_clean.groupby("Ship Mode")
    .agg({
        "Shipping Days": "mean",
        "Order ID": "nunique",
        "Sales": "sum"
    })
    .rename(columns={
        "Order ID": "Total Orders",
        "Shipping Days": "Average Shipping Days"
    })
)

shipping_analysis
```

```python
shipping_analysis["Average Shipping Days"].plot(
    kind="bar",
    title="Average Shipping Duration by Mode"
)

plt.ylabel("Days")
plt.xticks(rotation=20)
plt.show()
```
